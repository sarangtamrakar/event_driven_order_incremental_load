Moving source data to archieve after loading it to staging table

```python
source_volume_path = "/Volumes/test/default/order-data-volume/source/"
archieve_volume_path = "/Volumes/test/default/order-data-volume/archive/"


files = dbutils.fs.ls(source_volume_path)

for file in files:
    if not file.isDir():
        dbutils.fs.mv(file.path, archieve_volume_path + file.name)
```
