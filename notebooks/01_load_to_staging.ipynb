Read the data from source path dir(ammped volume)

```python
from pyspark.sql.functions import current_timestamp

source_path = "/Volumes/test/default/order-data-volume/source/"
staging_path = "/Volumes/test/default/order-data-volume/stage/delta/staging_order_data"

df = spark.read.option("header", True).csv(source_path)
df = df.withColumn("load_time", current_timestamp())
```

Write data to staging table in overwrite mode

```python
df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("test.default.staging_order_data")
```
